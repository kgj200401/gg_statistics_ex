```python
# data 불러오기
with open("./datas2/성인여성_키_데이터1.txt", "r") as f:
    data = f.read().split('\n')
    #print(list(map(float, data)))
    data = list(map(float, data))
    print(data) 
```

## 데이터의 평균과 표준편차 구하기

```python
from scipy import stats
import numpy as np

# 평균과 표준편차 계산
mean_value = np.mean(data)
std_dev = np.std(data, ddof=0)  # 모집단 표준편차(기본 설정)

mean_value, std_dev
```

## 가설 설정
- 귀무가설: 표본 데이터의 평균은 163과 같다.
- 대립가설: 표본 데이터의 평균은 163과 다르다.

## 정규성 검정
- 수집한 표본 데이터가 정규분포를 따르는지 검증한다.

## 방법 1 : KS-test

```python
# 정규성 검정, KS-test(Kolmogorov-Smirnov) 검정
# KS-test는 주어진 표본 데이터가 특정 이론적인 분포(예: 정규분포)를 따르는지 검정하는 방법

# 비교 대상이 되는 정규분포에 표본 데이터의 평균과 표준편차를 사용
ks_statistic, p_value = stats.kstest(
    data,
    lambda x: stats.norm.cdf(
        x,
        loc=np.mean(data),
        scale=np.std(data, ddof=0)
    )
)

ks_statistic, p_value
```

```python
import pandas as pd
import numpy as np
```

```python
# encoding = euc-kr로 되어있는데 이걸 지금 utf-8로 바꿔서 읽어야 함
df1 = pd.read_csv("./datas2/반별_점수_type1.csv", encoding = "utf-8")
# df1 = pd.read_csv("./datas2/반별_점수_type1.csv")
df1.head()
```

```python
# A반의 데이터만 추출
group_A = df1['점수'].loc[df1['반'] == 'A'].values
# B반의 데이터만 추출
group_B = df1['점수'].loc[df1['반'] == 'B'].values

print(group_A)
print(group_B)
```

```python
import numpy as np

# A반의 평균과 표준 편차
mean_A = np.mean(group_A)
std_A = np.std(group_A, ddof=1)  # 표본 표준 편차 (ddof=1)

# B반의 평균과 표준 편차
mean_B = np.mean(group_B)
std_B = np.std(group_B, ddof=1)  # 표본 표준 편차 (ddof=1)

# 결과 출력
print(f"A반 평균: {mean_A:.2f}, 표준 편차: {std_A:.2f}")
print(f"B반 평균: {mean_B:.2f}, 표준 편차: {std_B:.2f}")
```

## 정규성 검정
- 정규성 검정 가설수립
    - 영가설 : 두 그룹의 표본 데이터는 정규분포를 따른다.
    - 대립가설 : 두 그룹의 표본 데이터는 정규성을 따르지 않는다.

```python
# group_A와 group_B의 평균과 표준편차를 사용하여 정규성 검정 수행
print(stats.kstest(
  group_A, 
  stats.norm.cdf, 
  args=(np.mean(group_A), np.std(group_A, ddof=0))))

print(stats.kstest(
  group_B, 
  stats.norm.cdf, 
  args=(np.mean(group_B), np.std(group_B, ddof=0))))
```

### 해석

- p-value >= 0.05 : 모두 정규 분포를 따른다고 판단
- 독립표본 t검정 할 수 있음.

- 방법2 : Shapiro-Wilk 검정

```python
stat, p = stats.shapiro(group_A) 
print(f"group_A의 stat: {stat}, p: {p}")
stat, p = stats.shapiro(group_B) 
print(f"group_B의 stat: {stat}, p: {p}")
```
