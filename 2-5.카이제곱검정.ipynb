```python
import pandas as pd
```

```python
df = pd.read_csv("datas2/성별에따른만족도.csv", encoding='utf-8')
df.head()
```

```python
cross_table = pd.crosstab(df['성별'], df['만족도'])
cross_table
```

```python
#실제값 numpy array로 만들기
obs = cross_table.values
obs
```

# 카이제곱 검정
- 두 범주형 변수가 서로 독립적인지 검정한다.
    - 성별에 따른 만족도를 분석한다
        - 귀무가설(H0): 성별과 구매 여부는 서로 관련이 없다. 즉, 독립이다.
        - 대립가설(H1): 성별과 구매 여부는 서로 관련이 있다. 즉, 독립이 아니다.

```python
from scipy import stats
statistics, pvalue, dof, expected = stats.chi2_contingency(obs)
print(pvalue)
```

```python
statistics, dof, expected
```

```python
pd.DataFrame(expected, columns = cross_table.columns, index = cross_table.index)
```
